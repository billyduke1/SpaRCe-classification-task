```python
from drosolf import pns

pn = pns.pns() # loads receptor data and applies Olson model. returns a dataframe.
pn.describe()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 10))
plt.imshow(pn.to_numpy(), aspect="auto", cmap="viridis")
plt.colorbar(label="PN firing rate (Hz)")
plt.xlabel("Glomerulus")
plt.ylabel("Odour")
plt.title("Simulated PN responses to 110 odours")
plt.show()
```
